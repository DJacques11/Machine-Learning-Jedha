4. Separate the target variable "balance" into a Y object and the explanatory variables into a X object.

13. Compare each prediction made with the gini and entropy classifier, what proportion of predictions are identical?

You can either compare each prediction using a for loop over all observations' indices, otherwise you simply loop over the two prediction vectors using the ```zip``` function
You can figure out what ```zip``` does by running
```
A=[1,2,3]
B=["a","b","c"]

for a, b in zip(A,B):
  print(a,b)
```

The numbers appearing in the solution will not necessarily the ones you will obtain depending on random operations you ran on your data.

````python
# y_gini_pred = gini_classifier.predict(X_test)
# y_entropy_pred = decision_classifier.predict(X_test)

same_pred = 0
for i in range(len(y_gini_pred)):
    if y_gini_pred[i] == y_entropy_pred[i]:
        same_pred += 1

print(f"Same predictions between gini and entropy : {same_pred}")

diff_pred = 0
for i in range(len(y_gini_pred)):
    if y_gini_pred[i] != y_entropy_pred[i]:
       diff_pred += 1

print(f"Différent predictions between gini and entropy : {diff_pred}")
````

````python
same_pred_zip = 0
for gini_pred, entropy_pred in zip(y_gini_pred, y_entropy_pred):
    if gini_pred == entropy_pred:
        same_pred_zip += 1

print(same_pred_zip)

diff_pred_zip = 0
for gini_pred, entropy_pred in zip(y_gini_pred, y_entropy_pred):
    if gini_pred != entropy_pred:
        diff_pred_zip += 1

print(diff_pred_zip)
````

````python
mask = (y_entropy_pred == y_gini_pred)

count = (~mask).sum()

print(f"On the test set, there are {count} examples that have different predictions from Gini and Entropy")
print(f"Which corresponds to a percentage of {(np.round(count / len(y_gini_pred) * 100, 3))}% ")
````

## Fighting overfitting

18. How would you fight overfitting while still using decision tree classifiers? Try to find the set of hyper parameters that maximize the performances of the model on the test set. Did the model perform better ? Are there still ways to improve the model?

````python
new_decision_tree = DecisionTreeClassifier()
params ={
    'max_depth' : np.arange(1,10),
    'min_samples_split': [2, 5, 10, 20]
}

grid = GridSearchCV(estimator=new_decision_tree, param_grid=params, cv=10)

grid.fit(X_train, Y_train)
````

````python
print('Classification report on train set:')
print(classification_report(Y_train, grid.best_estimator_.predict(X_train)))
````

````python
print('Classification report on test set:')
print(classification_report(Y_test, grid.best_estimator_.predict(X_test)))
````
